```python
import time, torch
from diffusers import StableDiffusionPipeline

pipe = StableDiffusionPipeline.from_pretrained(
    "../models/sd15", torch_dtype=torch.float16, variant="fp16").to("cuda")

t = time.time()
img = pipe("a photo of a CEO", num_inference_steps=25).images[0]
print(f"{time.time() - t:.1f}초")
img.save("test.png")
img
```
